## Step 1 — Load Data

```python
from src.data_loader import load_data
df = load_data(data_dir='../data')
print(f'Shape: {df.shape}')
df.head()
```

## Step 2 — EDA

```python
from src.eda import explore_data, check_missing
explore_data(df)
check_missing(df)
```

```python
# Correlation heatmap
plt.figure(figsize=(12,9))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## Step 3 — Pre-processing

```python
from src.preprocessing import clean_data, split_features_target, get_train_test_split, scale_features

df_clean = clean_data(df.copy())
X, y     = split_features_target(df_clean)
X_train, X_test, y_train, y_test = get_train_test_split(X, y)
X_train_sc, X_test_sc, scaler    = scale_features(X_train, X_test)
print('Pre-processing done ✓')
```

## Step 4 — Train Models

```python
from src.train import build_models, train_and_evaluate

models     = build_models()
results_df = train_and_evaluate(models, X_train_sc, X_test_sc, y_train, y_test)
results_df
```

## Step 5 — Select & Save Best Model

```python
from src.train import select_best_model, save_model

best_name  = select_best_model(results_df)
best_model = models[best_name]
save_model(best_model, scaler, filename='best_model.pkl')
```

## Step 6 — Predict on a Sample Patient

```python
from src.predict import predict_from_input

sample = {
    'age':58,'sex':1,'cp':0,'trestbps':140,'chol':289,'fbs':0,
    'restecg':0,'thalach':140,'exang':1,'oldpeak':1.8,'slope':1,'ca':2,'thal':3
}
result = predict_from_input(sample, best_model, scaler)
print(result)
```
